## 3. What are the extra rows from?


```python
df.groupby('student_id').size().value_counts()
```

```python
sizes = df.groupby('student_id').size()
multi_ids = sizes[sizes > 4].index

multi = df[df.student_id.isin(multi_ids)].sort_values(['student_id', 'coverage'])
multi.head()
```

**Findings:** The extra rows are additional claims. This also confirms that the claims are not aggregated into 1 amount
